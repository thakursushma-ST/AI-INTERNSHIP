## Looping statements

```python
a=10
b=20
while(a<b):
    print(a)
    a+=1
```

```python
pass_word = "123"

new_password = input("Enter password: ")

while pass_word != new_password:
    print("Password does not match")
    new_password = input("Enter password again: ")

print("Password matched")
```

```python
n=10
for i in range(0,n):
    print(i)
```

```python
n=5
for i in range(0,n-1):
    for j in range(i+1,n):
        print(i,j)
```
